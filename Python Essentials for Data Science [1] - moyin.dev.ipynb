## 1️⃣ Python Lists
Lists are ordered, mutable collections used to store datasets and model outputs.

```python
scores = [70, 85, 90, 60]
scores
```

```python
# Accessing and modifying lists
scores[0]
scores.append(95)
scores[1] = 88
scores
```

## 3️⃣ Dictionaries
Dictionaries store data as key–value pairs. Very common in data science.

```python
student = {
    "name": "Alex",
    "age": 21,
    "score": 85
}
student
```

```python
# Dictionary operations
student["age"]
student["grade"] = "A"
student["score"] = 90
student
```

## 4️⃣ Functions
Functions help us reuse logic and keep code clean.

```python
def greet(name):
    return "Hello " + name

greet("Moyin")
```

```python
def check_result(score):
    cut_off = 60
    if score >= cut_off:
        return "Pass"
    else:
        return "Fail"

score = input("Enter your score: ")
result = check_result(int(score))
print("Your result is:", result)
```

## 6️⃣ Error Handling
Good programs handle errors gracefully.

```python
try:
    x = int(input("Enter a number: "))
    print(10 / x)
except ZeroDivisionError:
    print("You cannot divide by zero")
except ValueError:
    print("Invalid input")
```
