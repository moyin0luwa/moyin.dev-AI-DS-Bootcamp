## 1. Import Required Libraries

**What we're doing:** Loading the essential Python libraries we'll need for data analysis and visualization.

- **pandas (pd):** For data manipulation and analysis (think Excel on steroids)
- **numpy (np):** For numerical operations and mathematical functions
- **matplotlib.pyplot (plt):** For creating basic plots and charts
- **seaborn (sns):** For creating beautiful statistical visualizations

**Why it matters:** These are the industry-standard tools for data analysis in Python.

```python
# Import all necessary libraries for data analysis and visualization
import pandas as pd  # For working with data tables (DataFrames)
import numpy as np  # For numerical calculations
import matplotlib.pyplot as plt  # For basic plotting
import seaborn as sns  # For statistical visualizations

# Set the visual style for all plots to 'whitegrid' for clean, professional-looking charts
sns.set_style('whitegrid')
```

## 2. Load the Dataset

**What we're doing:** Reading our Excel file into a pandas DataFrame (a data table structure).

**Key concept:** 
- A DataFrame is like a spreadsheet in Python - it has rows and columns
- `.head()` shows us the first 5 rows to preview what our data looks like

**Tip:** Always inspect your data after loading to make sure it imported correctly!

```python
# Specify the path to our Excel file
file_path = 'student_performance - moyin.dev.xlsx'

# Read the Excel file into a pandas DataFrame
df = pd.read_excel(file_path)

# Display the first 5 rows to preview the data
# This helps us understand what columns we have and what the data looks like
df.head()
```

## 3. Explore the Dataset

**What we're doing:** Understanding the structure and quality of our data before analysis.

**Three key exploration steps:**
1. **`.info()`** - Shows column names, data types, and how many non-null values we have
2. **`.describe()`** - Provides statistical summary (mean, min, max, etc.) for numerical columns
3. **`.isnull().sum()`** - Counts missing values in each column

**Why this matters:** You need to know your data before you can analyze it properly!

```python
# Get a concise summary of the DataFrame
# Shows: column names, number of non-null entries, and data types
# Helps identify if any columns have missing data or wrong data types
df.info()
```

```python
# Generate descriptive statistics for numerical columns
# Shows: count, mean, std (standard deviation), min, 25%, 50%, 75%, max
# This gives us a quick sense of the distribution of our numerical data
df.describe()
```

```python
# Check for missing (null) values in each column
# Returns the count of missing values per column
# Zero means no missing data (which is ideal!)
df.isnull().sum()
```

## 4. Handling Missing Data

**What we're doing:** Dealing with any missing values in our dataset.

**Strategy:** We're filling missing attendance values with the average (mean) attendance.

**Why this approach:**
- Simple and commonly used method
- Preserves the overall distribution of the data
- Better than deleting rows with missing data

**Other options:** We could also use median, mode, or more advanced imputation techniques.

```python
# Fill any missing values in 'Attendance (%)' column with the mean (average) attendance
# This is called "mean imputation" - a common technique for handling missing numerical data
df['Attendance (%)'] = df['Attendance (%)'].fillna(df['Attendance (%)'].mean())

# Verify that we've successfully handled missing values by checking again
df.isnull().sum()
```

## 5. Basic Descriptive Statistics

**What we're doing:** Calculating key metrics to understand student performance.

**Three important questions we're answering:**
1. What's the overall average score across all students?
2. How does performance vary by course?
3. What's the pass rate for each course?

**Key learning:** The `.groupby()` function is incredibly powerful - it lets us split our data into groups and calculate statistics for each group.

```python
# Calculate the overall average score across ALL students and courses
# This gives us a single number representing the typical student performance
df['Score'].mean()
```

```python
# Calculate the average score for EACH course separately
# groupby('Course') splits the data by course
# Then .mean() calculates the average score for each group
# This helps us identify which courses are easier or harder
df.groupby('Course')['Score'].mean()
```

```python
# Create a new column 'Pass' that is True if score >= 50, False otherwise
# This creates a boolean (True/False) column we can use for calculations
df['Pass'] = df['Score'] >= 50

# Calculate the pass rate (%) for each course
# .mean() on a boolean column gives us the proportion of True values
# Multiplying by 100 converts it to a percentage
df.groupby('Course')['Pass'].mean() * 100
```

## 6. Filtering Data

**What we're doing:** Extracting specific subsets of data based on conditions.

**Key concept:** Boolean indexing - we use True/False conditions to filter rows.

**Syntax breakdown:**
- `df[df['Score'] > 85]` means "give me all rows where Score is greater than 85"
- `df[df['Term'] == 'Term 2']` means "give me all rows where Term equals 'Term 2'"

**Use cases:** Finding top performers, analyzing specific time periods, identifying at-risk students

```python
# Filter to show only high-performing students (scores above 85)
# The condition df['Score'] > 85 creates a True/False mask
# .head() limits output to first 5 rows for readability
df[df['Score'] > 85].head()
```

```python
# Filter to show only Term 2 results
# Note: Use == for equality comparison (not =, which is for assignment)
# This helps us analyze second term performance specifically
df[df['Term'] == 'Term 2'].head()
```

## 7. Grouping Data

**What we're doing:** Aggregating data by categories to identify patterns.

**Key insight:** `.groupby()` is one of the most powerful tools in pandas!

**What's happening:**
- We split the data into groups (by Course or by Term)
- We apply an aggregation function (like `.mean()`)
- We combine the results into a summary

**Real-world use:** This is how you answer questions like "Which department has the highest sales?" or "Which month had the best performance?"

```python
# Calculate average score by course (same as earlier - demonstrating the concept again)
# This is the "split-apply-combine" pattern:
# 1. Split data by Course
# 2. Apply mean() to each group
# 3. Combine results into a summary
df.groupby('Course')['Score'].mean()
```

```python
# Calculate average score by term
# This shows if students generally improved from Term 1 to Term 2
# We expect Term 2 to have a higher average if students are learning!
df.groupby('Term')['Score'].mean()
```

## 8. Improvement from Term 1 to Term 2

**What we're doing:** Identifying which students improved the most between terms.

**Key concept:** Pivot tables - reshaping data from "long" to "wide" format.

**Step-by-step:**
1. Create a pivot table with students as rows and terms as columns
2. Calculate improvement by subtracting Term 1 from Term 2
3. Sort to find the biggest improvements

**Why pivot tables:** They make it easy to compare values across different categories (like comparing Term 1 vs Term 2 for each student).

```python
# Create a pivot table showing each student's scores in Term 1 and Term 2
# Index: Student ID (rows)
# Columns: Term (Term 1, Term 2)
# Values: Score (averaged if a student has multiple scores per term)
pivot_scores = df.pivot_table(index='Student ID', columns='Term', values='Score')

# Calculate improvement by subtracting Term 1 score from Term 2 score
# Positive values = improvement, Negative values = decline
pivot_scores['Improvement'] = pivot_scores['Term 2'] - pivot_scores['Term 1']

# Sort by improvement in descending order and show top 5 students who improved most
# ascending=False means highest improvement first
pivot_scores.sort_values('Improvement', ascending=False).head()
```

## 9. Data Visualization

**What we're doing:** Creating charts and graphs to visualize patterns in our data.

**Why visualize:**
- Humans process visual information faster than tables of numbers
- Charts reveal patterns and outliers that might be hidden in raw data
- Visualizations make your analysis more compelling and easier to understand

**We'll create:**
- Bar charts (for comparing categories)
- Scatter plots (for relationships between variables)
- Histograms (for distributions)
- Box plots (for spread and outliers)
- Heatmaps (for correlations)

## Average Score per Course

**Chart type:** Bar chart

**What it shows:** Which courses have higher or lower average scores.

**How to read it:** Taller bars = higher average scores. This helps identify difficult courses.

```python
# Calculate average score per course and prepare for plotting
# reset_index() converts the groupby result back to a regular DataFrame
course_avg = df.groupby('Course')['Score'].mean().reset_index()

# Create a bar plot showing average scores by course
sns.barplot(x='Course', y='Score', data=course_avg)

# Rotate course names 45 degrees so they don't overlap
plt.xticks(rotation=45)

# Add a descriptive title
plt.title('Average Score per Course')

# Display the plot
plt.show()
```

## Average Score per Term

**Chart type:** Bar chart

**What it shows:** Whether students improved from Term 1 to Term 2.

**Expected pattern:** Term 2 should be higher if learning is happening!

```python
# Calculate average score per term and prepare for plotting
term_avg = df.groupby('Term')['Score'].mean().reset_index()

# Create a bar plot comparing Term 1 vs Term 2 performance
sns.barplot(x='Term', y='Score', data=term_avg)

# Add a descriptive title
plt.title('Average Score per Term')

# Display the plot
plt.show()
```

## Attendance vs Score

**Chart type:** Scatter plot

**What it shows:** The relationship between attendance and academic performance.

**How to read it:**
- Each dot represents one student record
- If dots trend upward from left to right → positive correlation (higher attendance = higher scores)
- If dots are scattered randomly → no clear relationship

```python
# Create a scatter plot to visualize relationship between attendance and scores
# Each point represents one student-course-term combination
# X-axis: Attendance percentage
# Y-axis: Score
sns.scatterplot(x='Attendance (%)', y='Score', data=df)

# Add a descriptive title
plt.title('Attendance vs Score')

# Display the plot
# Look for a positive trend: do students with better attendance tend to score higher?
plt.show()
```

## Score Distribution

**Chart type:** Histogram

**What it shows:** How scores are distributed across the entire dataset.

**How to read it:**
- Height of each bar = how many students fall in that score range
- Shape tells us if scores are normally distributed, skewed, or have multiple peaks
- Helps identify if most students are performing well or struggling

```python
# Create a histogram showing the distribution of all scores
# bins=10 divides the score range into 10 equal-width bins
# Each bar shows how many scores fall in that range
sns.histplot(df['Score'], bins=10)

# Add a descriptive title
plt.title('Score Distribution')

# Display the plot
# Look for the shape: is it bell-shaped (normal), skewed, or bimodal?
plt.show()
```

## Score Spread by Course

**Chart type:** Box plot

**What it shows:** The distribution and spread of scores for each course.

**How to read a box plot:**
- Box = middle 50% of scores (from 25th to 75th percentile)
- Line in box = median (50th percentile)
- Whiskers = extend to min/max values (within 1.5 × IQR)
- Dots = outliers (unusually high or low scores)

**Why it's useful:** Quickly compare variability and central tendency across courses.

```python
# Create a box plot showing score distribution for each course
# This helps us see not just the average, but the full spread of scores
sns.boxplot(x='Course', y='Score', data=df)

# Rotate course names for readability
plt.xticks(rotation=45)

# Add a descriptive title
plt.title('Score Spread by Course')

# Display the plot
# Compare box heights and positions to see which courses have more consistent scores
plt.show()
```

## Correlation Heatmap

**Chart type:** Heatmap

**What it shows:** How strongly different variables are related to each other.

**How to read it:**
- Values range from -1 to +1
- **+1** = perfect positive correlation (as one goes up, the other goes up)
- **-1** = perfect negative correlation (as one goes up, the other goes down)
- **0** = no correlation
- Darker colors typically indicate stronger correlations

**Key insight:** This helps us understand which factors most influence student scores.

```python
# Calculate correlation between Score, Attendance, and Assignments Submitted
# .corr() computes pairwise correlation of columns
# Create a heatmap to visualize these correlations
# annot=True displays the correlation values in each cell
sns.heatmap(df[['Score','Attendance (%)','Assignments Submitted']].corr(), annot=True)

# Add a descriptive title
plt.title('Correlation Matrix')

# Display the plot
# Strong positive correlations (close to 1) suggest that variable impacts scores
plt.show()
```
