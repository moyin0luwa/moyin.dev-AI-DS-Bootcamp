## 1. Arrays vs Lists
Python provides different ways to store data.  
The most common are **lists** and **arrays**.


```python

# Python List example
my_list = [1, 2, 3, 4, 5]
my_list
```

```python

# Lists can store different data types
mixed_list = [1, "hello", 3.5, True]
mixed_list
```


### NumPy Arrays
Arrays are optimized for numerical operations and are widely used in data science.


```python

import numpy as np

my_array = np.array([1, 2, 3, 4, 5])
my_array
```

```python

# Vectorized operation (very fast)
my_array * 2
```


## 2. Introduction to Pandas
Pandas is a powerful library for working with structured data.


```python

import pandas as pd
```


## 3. Pandas Series
A Series is a one-dimensional labeled array.


```python

series_example = pd.Series([10, 20, 30, 40])
series_example
```

```python

# Series with custom index
series_custom = pd.Series([100, 200, 300], index=["A", "B", "C"])
series_custom
```


## 4. Pandas DataFrames
A DataFrame is a two-dimensional table of data.


```python

data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [25, 30, 35],
    "Score": [85, 90, 95]
}

df = pd.DataFrame(data)
df
```


## 5. Loading CSV Files
CSV files are one of the most common data formats in data science.


```python

# Load a CSV file
# Replace the path with your own file if needed
df_csv = pd.read_csv("sample_data/california_housing_test.csv")
df_csv.head()
```


## 6. Exploring Your Data
Before analysis, always explore your data.


```python

df_csv.info()
```

```python

df_csv.describe()
```


## 7. Indexing and Slicing
Indexing allows you to select specific rows and columns.


```python

# Select a column
df_csv["median_house_value"].head()
```

```python

# Select multiple columns
df_csv[["median_income", "median_house_value"]].head()
```

```python

# Select rows using slicing
df_csv[0:5]
```

```python

# Conditional filtering
df_csv[df_csv["median_income"] > 5].head()
```
