# Traditional Logarithm Calculator

## Overview
This program calculates **log₁₀** (logarithm base 10) using the **traditional method** that mimics how logarithm tables were originally constructed in the 17th-19th centuries - before calculators existed!

---

## Understanding Logarithms

**What is log₁₀(x)?**

It answers the question: "What power do I raise 10 to, to get x?"

Examples:
- log₁₀(100) = 2 because 10² = 100
- log₁₀(1000) = 3 because 10³ = 1000
- log₁₀(2) ≈ 0.301 because 10^0.301 ≈ 2

---

## The Two-Part Structure

Every logarithm has two parts:

### 1. **Characteristic** (Integer Part)
The characteristic tells you the "scale" or "magnitude" of the number.

| Number | Characteristic | Why? |
|--------|---------------|------|
| 456 | 2 | Between 10² and 10³ |
| 45.6 | 1 | Between 10¹ and 10² |
| 4.56 | 0 | Between 10⁰ and 10¹ |
| 0.456 | -1 | Between 10⁻¹ and 10⁰ |

**How to find it:** Count how many times you divide (or multiply) by 10 to get a number between 1 and 10.

### 2. **Mantissa** (Decimal Part)
The mantissa is the precise decimal value. This is the tricky part!

---

## How The Algorithm Works

### STEP 1: Find the Characteristic
```
Start with: number = 456
While number >= 10:
    Divide by 10
    Count += 1

456 ÷ 10 = 45.6  (count = 1)
45.6 ÷ 10 = 4.56 (count = 2)
4.56 is between 1-10, STOP

Characteristic = 2
```

### STEP 2: Find the Mantissa (The Clever Part!)

This uses a brilliant mathematical trick:

**The Key Insight:**
If log₁₀(x) = 0.abcdef... (the mantissa)
Then x = 10^0.abcdef

When we raise both sides to the 10th power:
- x^10 = (10^0.abcdef)^10 = 10^(a.bcdef...)

Notice what happened? The decimal point shifted left by one position!
- The integer part "a" appeared
- We can extract this digit
- Repeat to get b, c, d, e, f...

**Example: Finding log₁₀(2)**

We know log₁₀(2) ≈ 0.301029...

Let's find it digit by digit:
```
Start: temp = 2

Round 1 - Finding first digit (0.3_____)
  temp^10 = 2^10 = 1024
  How many times can we divide by 10? 
    1024 ÷ 10 = 102.4
    102.4 ÷ 10 = 10.24
    10.24 ÷ 10 = 1.024 ✓
  Divided 3 times → First digit is 3
  Mantissa so far: 0.3
  Continue with temp = 1.024

Round 2 - Finding second digit (0.30____)
  temp^10 = 1.024^10 ≈ 1.267
  Can we divide by 10? No (already less than 10)
  Divided 0 times → Second digit is 0
  Mantissa so far: 0.30
  Continue with temp = 1.267

Round 3 - Finding third digit (0.301___)
  temp^10 = 1.267^10 ≈ 10.715
  How many times can we divide by 10?
    10.715 ÷ 10 = 1.0715 ✓
  Divided 1 time → Third digit is 1
  Mantissa so far: 0.301
  Continue with temp = 1.0715

...keep going for more precision...
```

Final answer: **log₁₀(2) ≈ 0.301029**

---

## Complete Example: log₁₀(456)

**Step 1: Characteristic**
- 456 ÷ 10 ÷ 10 = 4.56
- Divided 2 times
- **Characteristic = 2**

**Step 2: Mantissa**
- Now work with 4.56 (between 1 and 10)
- Use the digit-by-digit method...
- **Mantissa ≈ 0.659**

**Final Answer:**
log₁₀(456) = 2 + 0.659 = **2.659**

Check: 10^2.659 ≈ 456 ✓



````python
def log10_traditional(number, precision=6):
    """
    Calculate log base 10 using the traditional logarithm table method.
    This mimics how logarithm tables were originally constructed.
    
    Args:
        number: The number to find log10 of (must be positive)
        precision: Number of decimal places to calculate
    
    Returns:
        Approximate value of log10(number)
    """
    
    if number <= 0:
        raise ValueError("Logarithm is only defined for positive numbers")
    
    # ============================================
    # STEP 1: Find the CHARACTERISTIC (integer part)
    # ============================================
    # The characteristic tells us the "scale" of the number
    # Example: log10(456) = 2.something (because 10^2 = 100, 10^3 = 1000)
    
    characteristic = 0
    temp = number
    
    # If number >= 10, keep dividing by 10 and count
    while temp >= 10:
        temp /= 10
        characteristic += 1
    
    # If number < 1, keep multiplying by 10 and count (negative)
    while temp < 1:
        temp *= 10
        characteristic -= 1
    
    # Now temp is between 1 and 10
    # Example: 456 → temp = 4.56, characteristic = 2
    
    # ============================================
    # STEP 2: Find the MANTISSA (decimal part)
    # ============================================
    # This is the clever part! We find each decimal digit one at a time.
    # 
    # The key insight: If log10(x) = 0.abc...
    # Then: 10^0.abc = x
    # So: (10^0.abc)^10 = x^10 = 10^(a.bc...)
    # This shifts the decimal left by one position!
    
    mantissa = 0.0
    place = 1.0  # Keeps track of decimal place (0.1, 0.01, 0.001...)
    
    for _ in range(precision):
        # Raise to 10th power (this shifts the log's decimal left)
        temp **= 10
        
        # Count how many times we can divide by 10
        # This gives us the next digit of the log
        digit = 0
        while temp >= 10:
            temp /= 10
            digit += 1
        
        # Add this digit to mantissa at the correct decimal place
        place /= 10
        mantissa += digit * place
    
    # Combine characteristic and mantissa
    return characteristic + mantissa


# ========================================
# EXAMPLE WALKTHROUGH for log10(2)
# ========================================
print("=== Example: log10(2) ===\n")

print("Step 1: Find characteristic")
print("  2 is between 1 and 10")
print("  So characteristic = 0\n")

print("Step 2: Find mantissa (finding digits one by one)")
print("  Start with temp = 2")
print("  ")
print("  Round 1:")
print("    temp^10 = 2^10 = 1024")
print("    1024 ÷ 10 ÷ 10 ÷ 10 = 1.024 (divided 3 times)")
print("    First digit = 3 → mantissa = 0.3")
print("    temp is now 1.024")
print("  ")
print("  Round 2:")
print("    temp^10 = 1.024^10 ≈ 1.267")
print("    Can't divide by 10 (it's already < 10)")
print("    Second digit = 0 → mantissa = 0.30")
print("    temp is now 1.267")
print("  ")
print("  Round 3:")
print("    temp^10 = 1.267^10 ≈ 10.715")
print("    10.715 ÷ 10 = 1.0715 (divided 1 time)")
print("    Third digit = 1 → mantissa = 0.301")
print("    ...and so on\n")

print("Final result: log10(2) ≈ 0 + 0.301 = 0.301")
print("="*40)
print()

# ========================================
# RUN THE PROGRAM
# ========================================
num = float(input("Enter a number: "))
result = log10_traditional(num)
print(f"\nlog10({num}) ≈ {result:.6f}")

# Verify with actual value
import math
actual = math.log10(num)
error = abs(result - actual)
print(f"Actual value: {actual:.6f}")
print(f"Error: {error:.6f}")
````
